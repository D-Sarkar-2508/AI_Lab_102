```python
import networkx as nx
import matplotlib.pyplot as plt

# Create directed graph
G = nx.DiGraph()
# Animal relationships
G.add_edge("Animal", "Mammal", relation="is-a")
G.add_edge("Animal", "Bird", relation="is-a")

# Mammal relationships
G.add_edge("Mammal", "Dog", relation="is-a")
G.add_edge("Mammal", "Cat", relation="is-a")

# Properties
G.add_edge("Mammal", "Fur", relation="has")
G.add_edge("Bird", "Feathers", relation="has")

# Abilities
G.add_edge("Dog", "Barks", relation="can")
G.add_edge("Cat", "Meows", relation="can")
G.add_edge("Bird", "Flies", relation="can")

pos = {
    "Animal": (0, 5),

    "Mammal": (-2.5, 3.8),
    "Bird": (2.5, 3.8),

    "Fur": (-4.5, 3.8),
    "Feathers": (4.5, 3.8),

    "Dog": (-3.2, 2.2),
    "Cat": (-1.2, 2.2),

    "Barks": (-3.2, 0.5),
    "Meows": (-1.2, 0.5),

    "Flies": (2.5, 1.5)
}
plt.figure(figsize=(12, 8))

nx.draw_networkx_nodes(
    G,
    pos,
    node_size=2600,
    node_color="lightblue",
    edgecolors="black",
    linewidths=1.5
)

# Node labels
nx.draw_networkx_labels(
    G,
    pos,
    font_size=11,
    font_weight="bold"
)
nx.draw_networkx_edges(
    G,
    pos,
    arrows=True,
    arrowsize=20,
    arrowstyle="-|>",
    width=1.8,
    edge_color="gray",
    node_size=2600
)
edge_labels = nx.get_edge_attributes(G, "relation")

nx.draw_networkx_edge_labels(
    G,
    pos,
    edge_labels=edge_labels,
    font_size=10,
    font_weight="bold",
    label_pos=0.5,
    bbox=dict(
        facecolor="white",
        edgecolor="none",
        alpha=0.8
    )
)
plt.title(
    "Semantic Network of Animals",
    fontsize=18,
    fontweight="bold",
    pad=20
)

plt.axis("off")
plt.tight_layout()
plt.show()
```
